```python
#  Import Library
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# Load Dataset
data = pd.read_csv('../review_gopay_newest_sort.csv')
print("Jumlah data:", len(data))# load dataset

text_column = "content"  # sesuaikan dengan nama kolom
print("Kolom content:", data[text_column].head(9))
```

```python
# Sentence Segmentation

# Memecah teks menjadi kalimat:
text = "Saya belajar NLP. NLP sangat menarik. Saya ingin menjadi NLP Engineer."

sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)
```

```python
# Word Tokenization
from nltk.tokenize import word_tokenize

text = "Saya belajar Natural Language Processing."

tokens = word_tokenize(text)

print(tokens)
```

```python
# Character Tokenization
text = "NLP"

characters = list(text)

print(characters)
```

```python
# Tokenisasi Dataset

# Menerapkan word tokenization ke seluruh dataset:
data["tokens"] = data[text_column].astype(str).apply(word_tokenize)

data[[text_column, "tokens"]].head()

# Menghitung jumlah token per dokumen:
data["jumlah_token"] = data["tokens"].apply(len)
data["jumlah_token"].describe()
```

```python
# Membandingkan Kata dan Token

# Hitung jumlah kata dengan split sederhana:
data["jumlah_kata"] = data[text_column].astype(str).apply(
    lambda x: len(x.split())
)

# Bandingkan:
data[["jumlah_kata", "jumlah_token"]].describe()
```

```python
# Eksperimen Tokenisasi
texts = [
    "Saya suka produk ini.",
    "Bagus banget!!!",
    "Tidak bagus...",
    "Harga Rp50.000.",
    "email@example.com",
    "https://example.com",
    "yg bagus bgt sih",
    "aku udah beli nih kak"
]

for text in texts:
    print("TEXT :", text)
    print("TOKEN:", word_tokenize(text))
    print()
```

```python
# Perbandingan Tokenizer
import spacy
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

# Contoh teks
teks_list = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000"
]

# Membandingkan hasil tokenisasi
hasil = []

for teks in teks_list:
    nltk_tokens = word_tokenize(teks)
    split_tokens = teks.split()

    hasil.append({
        "Teks": teks,
        "NLTK word_tokenize": str(nltk_tokens),
        "split()": str(split_tokens)
    })

tabel = pd.DataFrame(hasil)
print(tabel.to_string(index=False))
```
